# Dog Brain registration project

### 1. Pre-processing
- goal: to make histology slides look like the MRI for easier matching
- <— **initial input**: raw slides
- process:
    - downsample to MRI resolution,
    - colour adjustment (grey scale, correct for the blue tint, adjust contrast)
    - optional smoothing (try Gaussian or median filter)
- —> **output**: pre-processed slide

### 2. Locate slide within the 3D MRI? ----- MANUAL FOR NOW
- <— **input**: pre-processed-slide, 3D MRI
- process:
    - image comparison (intensity, or feature location?) to determine which 2D plane is the most alike to the processed histology slide?
    - several initialisations to avoid falling into a local min during gradient descent?
    - likely to be computationally expensive, haven't found how they do it in literature, unless they have manually placed key features on both reference and moving images.
- —> **output** a 2D plane where to focus the registration effort

### 3. Registration:
- <— **input**: preprocessed slide and corresponding 2D plane in 3D MRI
- process:
    - find out the transformation matrix (translation, rotation, shear)
    - use existing packages: homography,  light glue, FSL, warpy
- —> **output**: transformation matrix

### 4. Apply transformation matrix:
- !! blue stains need to be visible in the final registered image
- <— **input**: original histology slide, selected 2D plane from MRI, transformation matrix (and whatever rotation in step 1 we might have made to facilitate the registration)
- —> **output**: original slide within the MRI structure, and check it overlays nicely + blue stain visible
